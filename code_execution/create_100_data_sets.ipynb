```python
from pathlib import Path
import random
import shutil
import pandas as pd


SOURCE_IMAGES_DIR = Path("data/niftis")
SOURCE_SUBMISSION_PATH = Path("data/submission_format.csv")

TARGET_ROOT = Path("data/test_100")
TARGET_IMAGES_DIR = TARGET_ROOT / "niftis"
TARGET_SUBMISSION_PATH = TARGET_ROOT / "submission_format.csv"

N_IMAGES = 100
RANDOM_SEED = 42


# Wczytanie oryginalnego submission_format.csv.
submission_df = pd.read_csv(SOURCE_SUBMISSION_PATH)

if len(submission_df) < N_IMAGES:
    raise ValueError(
        f"CSV zawiera tylko {len(submission_df)} rekordów."
    )

if "uid" not in submission_df.columns:
    raise ValueError(
        f"Brak kolumny 'uid'. Kolumny: {submission_df.columns.tolist()}"
    )


# Losowanie 100 wierszy w sposób powtarzalny.
random_generator = random.Random(RANDOM_SEED)

selected_indices = random_generator.sample(
    list(submission_df.index),
    N_IMAGES,
)

# Sortowanie indeksów zachowuje kolejność z oryginalnego CSV.
selected_indices.sort()

submission_100_df = (
    submission_df
    .loc[selected_indices]
    .reset_index(drop=True)
)


# Utworzenie katalogu docelowego.
TARGET_IMAGES_DIR.mkdir(
    parents=True,
    exist_ok=True,
)


# Kopiowanie obrazów odpowiadających wybranym UID.
for index, uid in enumerate(
    submission_100_df["uid"],
    start=1,
):
    possible_paths = [
        SOURCE_IMAGES_DIR / f"{uid}.nii.gz",
        SOURCE_IMAGES_DIR / f"{uid}.nii",
    ]

    source_path = next(
        (
            path
            for path in possible_paths
            if path.exists()
        ),
        None,
    )

    if source_path is None:
        raise FileNotFoundError(
            f"Nie znaleziono obrazu dla UID: {uid}"
        )

    target_path = TARGET_IMAGES_DIR / source_path.name

    shutil.copy2(
        source_path,
        target_path,
    )

    print(
        f"{index:3d}/{N_IMAGES} | {uid}",
        flush=True,
    )


# Zapis dopasowanego submission_format.csv.
submission_100_df.to_csv(
    TARGET_SUBMISSION_PATH,
    index=False,
)


print("\nGotowe")
print(f"Obrazy: {TARGET_IMAGES_DIR.resolve()}")
print(f"CSV:    {TARGET_SUBMISSION_PATH.resolve()}")
print(f"Liczba rekordów: {len(submission_100_df)}")
```
