```python
from pathlib import Path
import pandas as pd
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
```

```python
PROJECT_DIR = Path.cwd().parent
LABELS_PATH = PROJECT_DIR / "data" / "niftis" / "train_labels_JNDlMjr.csv"

labels = pd.read_csv(LABELS_PATH)
print(labels["is_pathologic"].value_counts())
print(labels["is_pathologic"].value_counts(normalize=True) * 100)
```

```python
NIFTI_DIR = PROJECT_DIR / "data" / "niftis"

files_uid = {f.stem.replace(".nii", "") for f in NIFTI_DIR.glob("*.nii.gz")}
labels_uid = set(labels["uid"])

print(f"Liczba plików: {len(files_uid)}")
print(f"Liczba etykiet: {len(labels_uid)}")
print(f"Brakujących plików: {labels_uid - files_uid}")
```

```python
files = sorted(NIFTI_DIR.glob("*.nii.gz"))
img = nib.load(files[0])
print(type(img))

volume = img.get_fdata(dtype=float)

print(f"Shape: {volume.shape}")
print(f"Dtype: {volume.dtype}")
print(f"Spacing: {img.header.get_zooms()}")
print(f"Min: {volume.min()}")
print(f"Max: {volume.max()}")
print(f"Mean: {volume.mean():.2f}")
print(f"Std: {volume.std():.2f}")
```

```python
print("Typ zapisany w pliku:", img.get_data_dtype())
print("Typ tablicy po get_fdata:", volume.dtype)
```
