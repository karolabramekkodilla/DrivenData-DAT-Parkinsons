```python
from pathlib import Path
import shutil
import pandas as pd


CODE_DIR = Path(
    r"C:\Users\abram\PycharmProjects"
    r"\PythonProject2\code_execution"
)

SOURCE_DATA_DIR = CODE_DIR / "data"
SOURCE_NIFTI_DIR = SOURCE_DATA_DIR / "niftis"

SMOKE_DATA_DIR = CODE_DIR / "data_smoke"
SMOKE_NIFTI_DIR = SMOKE_DATA_DIR / "niftis"

SMOKE_NIFTI_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

submission = pd.read_csv(
    SOURCE_DATA_DIR
    / "submission_format.csv"
)

# Wybieramy tylko UID, dla których obraz faktycznie istnieje.
available_mask = submission["uid"].apply(
    lambda uid: (
        SOURCE_NIFTI_DIR
        / f"{uid}.nii.gz"
    ).exists()
)

available_submission = submission.loc[
    available_mask
].copy()

print(
    "Dostępne obrazy:",
    len(available_submission),
)

# Krótki test, np. 10 obrazów.
smoke_submission = (
    available_submission
    .head(10)
    .copy()
)

if smoke_submission.empty:
    raise FileNotFoundError(
        f"Brak pasujących obrazów w "
        f"{SOURCE_NIFTI_DIR}"
    )

for uid in smoke_submission["uid"]:
    source_path = (
        SOURCE_NIFTI_DIR
        / f"{uid}.nii.gz"
    )

    target_path = (
        SMOKE_NIFTI_DIR
        / f"{uid}.nii.gz"
    )

    shutil.copy2(
        source_path,
        target_path,
    )

smoke_submission.to_csv(
    SMOKE_DATA_DIR
    / "submission_format.csv",
    index=False,
)

print("Utworzono:", SMOKE_DATA_DIR)
print(smoke_submission)
```
