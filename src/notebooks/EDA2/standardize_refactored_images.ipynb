```python
import nibabel as nib
import matplotlib.pyplot as plt
from pathlib import Path
import pandas as pd
from ipywidgets import interact, IntSlider
from notebooks.create_mri_view import create_mri_view
from resources.standarize_image import standardize_image
```

```python
PROJECT_DIR = Path.cwd().parents[1]
OUTPUTS_DIR = PROJECT_DIR / "outputs"
SCAN_CENTRES_PATH = OUTPUTS_DIR / "scan_centres.csv"
CUT_IMAGES_DIR = PROJECT_DIR / "outputs" / "cut_images"
scan_centres = pd.read_csv(SCAN_CENTRES_PATH)
uid_to_show = ["3m9iqbq9","01nouhtc", "wnnhj7xv","gv0e38da","pnsm43d8"]
```

```python
for uid in uid_to_show:

    file = CUT_IMAGES_DIR / f"{uid}.nii.gz"

    target = scan_centres.loc[
        scan_centres["uid"] == uid,
        "pathological"
    ].iloc[0]

    diagnosis = (
        "Pathological"
        if target == 1
        else "Not pathological"
    )

    volume = nib.load(file).get_fdata()
    image = nib.load(file)

    image = standardize_image(image, percent_top=0.02)
    image = image.get_fdata()

    print("Plik:", file.name)
    print("Shape:", volume.shape)
    print("Min:", volume.min())
    print("Max:", volume.max())

    create_mri_view(
        uid=uid,
        volume=volume,
        diagnosis=diagnosis
    )
    print("Shape:", image.shape)
    print("Min:", image.min())
    print("Max:", image.max())
    create_mri_view(
        uid=uid,
        volume=volume,
        diagnosis=diagnosis
    )
```
