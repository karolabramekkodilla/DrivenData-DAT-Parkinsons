```python
voxel_label_stats = (
    data.groupby(
        ["x_voxel_size", "y_voxel_size", "z_voxel_size"]
    )
    .agg(
        count=("pathological", "size"),
        pathological_count=("pathological", "sum"),
        pathological_rate=("pathological", "mean"),
        uids=("uid", list)
    )
    .reset_index()
    .sort_values("count", ascending=False)
)

voxel_label_stats
```
