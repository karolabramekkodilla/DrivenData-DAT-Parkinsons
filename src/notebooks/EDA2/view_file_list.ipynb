```python
selected_uids = ['9vjuhbp9', '9jrd4hqf', '3hzcd3ge', 'g7f9fzja', 'bsepr1b2', '948ju63l', '5r8vo4kv', 'trd83x9a', 'vyb4xz3m','vzme9wt0', 'ed8u8noz', 'oi9wm1sa', 'v9cng8ta', 'na6u48gw', 'ghm56vu6', 'z5gf3bbt', '873lkmgs', 'gom6736z', '4dsxlsut', '11zsb8lv', 'hxa7hy7a', 'i33993n4', 'wh50jyoo', 'l7jy7otk', '3iwpfvb7', '1ghlu8m8', '7mhxofkc', 'ggxp5cit', 'uf8cyotw', '31eiwucx', 'wm9q7ftm', 'fwg895na', '2fqztio4', 'm7pp0pnn', 'lo0hhgh5','uz2z0kdn', '839ag77q', 'vqa7izq2', 'nmnr06dj', 'zbc7qk4k', '287gt76i', 'q10f7ia4', 'xsy4vy32', 'i3pj7k99', 'p3frb7ei']
```

```python
PROJECT_DIR = Path(r"C:\Users\abram\PycharmProjects\DrivenData_DAT_Parkinsons")
NIFTI_DIR = PROJECT_DIR / "data" / "niftis"
filter_width = 1
LABELS_PATH = PROJECT_DIR / "data" / "niftis" / "train_labels_JNDlMjr.csv"
labels = pd.read_csv(LABELS_PATH)

```

```python
from ipywidgets import IntSlider

for uid in selected_uids:
    FILE_PATH = NIFTI_DIR / f"{uid}.nii.gz"

    original_img = nib.load(FILE_PATH)
    original = original_img.get_fdata(dtype=np.float32)

    cropped, _ = cut_off_black_segments(
        path=FILE_PATH,
        threshold=0,
        margin=1,
    )

    x_max, y_max, z_max = cropped.shape

    filter_3d_x = np.ones(
        (filter_width, y_max, z_max)
    )

    filter_sum_3d_x = np.zeros(
        x_max - filter_width + 1
    )

    for x_pos in range(x_max - filter_width + 1):
        area = cropped[
            x_pos:x_pos + filter_width,
            :,
            :
        ]

        filter_sum_3d_x[x_pos] = np.sum(
            area * filter_3d_x
        )

    # Odczyt etykiety pathological dla danego UID
    label_rows = labels.loc[
        labels["uid"] == uid,
        "is_pathologic"
    ]

    if label_rows.empty:
        pathological = "brak etykiety"
    else:
        pathological = int(label_rows.iloc[0])

    z_slider = IntSlider(
        value=z_max // 2,
        min=0,
        max=z_max - 1,
        step=1,
        description="Warstwa Z:",
        continuous_update=False,
    )

    output = Output()

    def draw_scan(change=None,
                  uid=uid,
                  cropped=cropped,
                  filter_sum_3d_x=filter_sum_3d_x,
                  pathological=pathological,
                  z_slider=z_slider):

        z = z_slider.value

        with output:
            output.clear_output(wait=True)

            fig, axes = plt.subplots(
                1,
                2,
                figsize=(15, 6)
            )

            # Wykres filtra
            axes[0].plot(
                np.arange(len(filter_sum_3d_x)),
                filter_sum_3d_x
            )

            axes[0].set_title(
                f"{uid}\n"
                f"pathological = {pathological}\n"
                f"Filtr 3D o szerokości {filter_width}"
            )

            axes[0].set_xlabel(
                "Pozycja początku filtra X"
            )
            axes[0].set_ylabel(
                "Suma intensywności"
            )
            axes[0].grid(True)

            # Widok z góry
            axes[1].imshow(
                cropped[:, :, z].T,
                cmap="gray",
                origin="lower"
            )

            axes[1].set_title(
                f"{uid}\n"
                f"pathological = {pathological}\n"
                f"Widok z góry: z = {z}"
            )

            axes[1].set_xlabel("X")
            axes[1].set_ylabel("Y")

            plt.tight_layout()
            plt.show()

    z_slider.observe(
        draw_scan,
        names="value"
    )

    display(z_slider, output)
    draw_scan()
```
